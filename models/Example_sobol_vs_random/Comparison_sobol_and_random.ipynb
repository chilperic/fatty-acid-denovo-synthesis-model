```python
from scipy.stats import qmc
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.colors import ListedColormap
import numpy as np
from pathlib import Path
import scienceplots
#plt.style.use("default")
plt.style.use('science')


#plt.rcParams["font.family"] = "serif"
# plt.rcParams["font.weight"] = "bold"
# plt.rcParams["axes.labelweight"] = "bold"
# plt.rcParams['axes.titlepad'] = 23
plt.rcParams.update({'figure.dpi': '300'})
plt.rcParams.update({
    "font.family": "serif",   # specify font family here
    "font.serif": ["Times"],  # specify font here
    "font.size":11}) 

colors = plt.rcParams["axes.prop_cycle"].by_key()["color"]
cols = [f"{color}" for color in colors]
RESULTS_DIR = Path("results")
RESULTS_DIR.mkdir(exist_ok=True)
```

```python

sns.set(style = "darkgrid")
```

```python
def generate_sobol_sequence(n, k):
    """
    n: number of sample
    m:power
    Generate a Sobol sequence in the given dimension with n points.
    """
    sampler= qmc.Sobol(d=n, scramble=False)
    sample = sampler.random_base2(m=k)
    return sample
```

```python
n, m = 3,8
X= generate_sobol_sequence(n, m)
```

```python
df = pd.DataFrame(X, columns=['x','y','z'])
df
```

```python
x1,y1,z1 = df["x"], df["y"], df["z"]
```

```python
l=2**8
x2, y2,z2 = sample = np.random.random(l), np.random.random(l), np.random.random(l)
```

```python
# Visualize the Sobol sequence using scatter3D
cmap = ListedColormap(sns.color_palette("husl", 256).as_hex())
fig = plt.figure(figsize=(3,3))
ax = Axes3D(fig, auto_add_to_figure=True,)
fig.add_axes(ax)

ax.scatter3D(x1, y1, z1, s=10, marker='.', cmap=cmap, alpha=1)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
fig.patch.set_alpha(0.0)
fig.tight_layout()
plt.show()
```

```python
qmc.discrepancy(X)
```

```python
cmap = ListedColormap(sns.color_palette("husl", 256).as_hex())
fig = plt.figure(figsize=plt.figaspect(0.5))
ax = fig.add_subplot(1, 2, 1, projection='3d')

ax.scatter3D(x1, y1, z1, s=10, marker='.', cmap=cmap, alpha=1)
# ax.scatter3D(x2, y2, z2, s=10, marker='.', cmap=cmap, alpha=1)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.set_title('Sobol sampling')
ax = fig.add_subplot(1, 2, 2, projection='3d')
#ax = Axes3D(fig, auto_add_to_figure=True,)
fig.add_axes(ax)

ax.scatter3D(x2, y2, z2, s=10, marker='.', cmap=cmap, alpha=1)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.set_title('random sampling')
fig.patch.set_alpha(0.0)
plt.tight_layout()
plt.show()


```
